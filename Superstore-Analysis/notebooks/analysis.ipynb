```python
import pandas as pd

df = pd.read_csv(r"D:\Superstore-Analysis\data\SampleSuperstore.csv")

df = df.drop_duplicates()
df = df.dropna()
df.columns = df.columns.str.strip()

print("Total Sales:", df['Sales'].sum())
print("Total Profit:", df['Profit'].sum())

print("\nSales by Category:")
print(df.groupby('Category')['Sales'].sum())

print("\nProfit by Region:")
print(df.groupby('Region')['Profit'].sum())

print("\nSegment Performance:")
print(df.groupby('Segment')[['Sales', 'Profit']].sum())

df['Profit Ratio'] = df['Profit'] / df['Sales']

loss = df[df['Profit'] < 0]

print("\nLoss Making Records:")
print(loss[['Category', 'Sub-Category', 'Profit']].head())

print("\nDiscount vs Profit:")
print(df.groupby('Discount')['Profit'].mean())
```
