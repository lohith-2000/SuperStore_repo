```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
df = pd.read_csv(r"D:\Superstore-Analysis\data\SampleSuperstore.csv")
sns.barplot(x='Category', y='Sales', data=df)
plt.title("Sales by Category")
plt.savefig("category_sales.png")
plt.show()

sns.barplot(x='Region', y='Profit', data=df)
plt.title("Profit by Region")
plt.savefig("region_profit.png")
plt.show()

sns.scatterplot(x='Discount', y='Profit', data=df)
plt.title("Discount vs Profit")
plt.savefig("discount_profit.png")
plt.show()

top_subcat = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).head(10)
top_subcat.plot(kind='barh')
plt.title("Top Sub-Categories by Sales")
plt.savefig("top_subcategories.png")
plt.show()
```
