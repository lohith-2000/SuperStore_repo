```python
import pandas as pd
df = pd.read_csv(r"D:\Superstore-Analysis\data\SampleSuperstore.csv")
with open("report.txt", "w") as f:
    f.write("Total Sales: " + str(df['Sales'].sum()) + "\n")
    f.write("Total Profit: " + str(df['Profit'].sum()) + "\n\n")

    f.write("Sales by Category:\n")
    f.write(str(df.groupby('Category')['Sales'].sum()) + "\n\n")

    f.write("Profit by Region:\n")
    f.write(str(df.groupby('Region')['Profit'].sum()) + "\n\n")

    f.write("Segment Performance:\n")
    f.write(str(df.groupby('Segment')[['Sales', 'Profit']].sum()) + "\n")
```
